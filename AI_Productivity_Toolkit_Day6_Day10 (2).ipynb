## 1. Introduction to Prompt Engineering

Prompt engineering is the process of designing clear and structured instructions for an AI model to obtain useful and relevant outputs.

A good prompt can include:
- Role
- Context
- Task
- Instructions
- Constraints
- Examples
- Output format

## 2. Zero-Shot Prompting

Zero-shot prompting means asking an AI model to perform a task without providing examples.

### Example

Classify the following review as Positive, Negative, or Neutral:

"The product arrived quickly and works perfectly."

Expected Output:

Positive

### Observation

The AI can perform the classification without being given any examples. This demonstrates zero-shot prompting.

## 3. Few-Shot Prompting

Few-shot prompting provides the AI model with a small number of examples before asking it to perform a new task.

### Example

Classify each review as Positive or Negative.

Example 1:
Review: "The product is excellent and very useful."
Classification: Positive

Example 2:
Review: "The product stopped working after one day."
Classification: Negative

Now classify:

Review: "The product is easy to use and works well."
Classification:

### Expected Output

Positive

### Observation

Few-shot prompting provides examples of the desired input-output pattern. These examples help the AI understand how the task should be performed and how the output should be formatted.

# Day 7 — Roles, Context, Instructions and Constraints

## 4. Role-Based Prompting

Role-based prompting gives the AI a specific role or expertise so that the response can be more focused and relevant.

### Example 1: Python Teacher

You are an experienced Python programming teacher.

Explain Python lists to a beginner.
Use simple language and provide 3 examples.

### Example 2: Technical Interviewer

You are a technical interviewer.

Ask me 5 beginner-level Python interview questions.
Do not provide the answers immediately.

### Example 3: Career Mentor

You are a career mentor.

Create a 7-day beginner study plan for learning Python
for technical placement preparation.
Keep the plan practical and beginner-friendly.

### Observation

Giving the AI a specific role helps define the type of response, tone and expertise expected from the model.

## 5. Context, Instructions and Constraints

### Example

Role:
You are a Python teacher.

Context:
The learner is a beginner and has never studied Python deeply.

Task:
Explain Python functions.

Instructions:
Explain the concept step by step and provide simple examples.

Constraints:
- Use simple English.
- Give 2 examples.
- Keep the explanation under 300 words.

Output Format:
Use headings and code blocks.

### Observation

Adding a role, context, task, instructions, constraints and output format makes the prompt clearer and helps the AI produce a more controlled and relevant response.

# Day 8 — Structured Outputs

## 6. Structured JSON Output

Structured output means asking the AI to return information in a specific format, such as JSON.

### Example

Extract the following information from the text and return ONLY valid JSON.

Text:

"Rahul is a Computer Science student. He knows Python and SQL. He is interested in Artificial Intelligence."

Required JSON format:

{
  "name": "",
  "education": "",
  "skills": [],
  "interest": ""
}

### Expected Output

{
  "name": "Rahul",
  "education": "Computer Science student",
  "skills": ["Python", "SQL"],
  "interest": "Artificial Intelligence"
}

### Observation

Structured JSON output organizes information in a consistent format. This makes AI-generated information easier for applications and software systems to process.

## 7. Summarization

Summarization prompts ask the AI to reduce a longer piece of information into a shorter version while keeping the important points.

### Example Prompt

Summarize the following paragraph in exactly 3 bullet points.

Text:

Artificial Intelligence is increasingly being used in healthcare, education, finance, transportation and customer service. Machine learning allows systems to learn patterns from data and make predictions. Generative AI can create text, images, code and other content. Responsible AI requires attention to accuracy, privacy, security and fairness.

Requirements:
- Exactly 3 bullet points
- Use simple English
- Preserve the important information

### Expected Output

- AI is used in many fields such as healthcare, education, finance and transportation.
- Machine learning learns patterns from data, while Generative AI can create different types of content.
- Responsible AI requires attention to accuracy, privacy, security and fairness.

## 8. Classification

Classification prompts ask the AI to assign information to a predefined category.

### Example Prompt

Classify the following message into one of these categories:

Categories:
- Internship
- Education
- Personal
- Technical

Message:

"Congratulations! You have been selected for the Machine Learning Internship program."

Return only the category.

### Expected Output

Internship

## 9. Content Generation

Content generation prompts ask the AI to create new content according to specific requirements.

### Example Prompt

Generate a professional email requesting an internship opportunity.

Requirements:
- Professional tone
- Maximum 150 words
- Mention interest in AI and Machine Learning
- Mention that I am a second-year B.Tech CSE AI & ML student
- Include a clear subject line

### Observation

Different prompt structures can be used for different AI tasks. Summarization reduces information, classification assigns categories, and generation creates new content based on the given instructions.

# Day 9 — AI-Assisted Coding and Debugging

## 10. AI-Assisted Coding

AI can assist programmers by generating code, explaining code and suggesting improvements.

### Example Prompt

You are a Python tutor.

Write a Python program that:
1. Takes a list of numbers.
2. Finds the smallest number.
3. Finds the second smallest number.
4. Displays both results.

Use beginner-friendly Python.
Explain the code after providing it.

```python
numbers = [10, 4, 8, 2, 6]

sorted_numbers = sorted(numbers)

smallest = sorted_numbers[0]
second_smallest = sorted_numbers[1]

print("Smallest:", smallest)
print("Second smallest:", second_smallest)
```

## 11. AI-Assisted Debugging

AI can help identify programming errors, explain why they occur and provide corrected code.

### Example Prompt

You are a Python debugging assistant.

Find the error in the following code.

Explain:
1. What is wrong?
2. Why is it wrong?
3. Provide the corrected code.

Code:

numbers = [10, 4, 8, 2, 6]

smallest = numbers[0]

for num in numbers:
    if num < smallest
        smallest = num

print("Smallest:", smallest)

### Error Identified

The `if` statement is missing a colon (`:`) at the end of the condition.

Incorrect:

if num < smallest

Correct:

if num < smallest:

```python
numbers = [10, 4, 8, 2, 6]

smallest = numbers[0]

for num in numbers:
    if num < smallest:
        smallest = num

print("Smallest:", smallest)
```

### Observation

AI-assisted coding can help generate beginner-friendly programs and explain programming concepts. AI-assisted debugging can identify syntax errors and suggest corrections. The generated code should always be tested and verified by the programmer.

## 12. Prompt Evaluation

Prompt evaluation means testing prompts to check whether they produce clear, relevant, accurate and consistent results.

### Evaluation Criteria

Prompts are evaluated using:

- Clarity
- Relevance
- Correctness
- Instruction following
- Output format
- Consistency

### Example: Improving a Prompt

#### Basic Prompt

Write a resume.

You are a professional resume assistant.

Create a one-page beginner resume for a second-year B.Tech CSE AI & ML student.

- Career objective
